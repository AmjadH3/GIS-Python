3. Write a Python program using a for loop to count how many even numbers exist between two user-provided
integers (inclusive).

```python
start = int(input("Enter a start number"))
end = int(input("Enter an end number"))

countEven = 0

for num in range (start, end + 1):
    if num % 2 == 0:
        countEven += 1
print (f"There are {countEven} even numbers between {start} and {end}")
```

4. Write a program that calculates the factorial of a given number using a while loop.

```python
num = int(input("Enter a number to calculate its factorial: "))
factorial = 1
counter = num
while counter > 0:
    factorial *= counter
    counter -= 1
print (f"The factorial of {num} is {factorial}")
```

6. Write a Python program using a while loop to print the Fibonacci series up to a user-specified limit.
(Example: If input is 10, output should be 0, 1, 1, 2, 3, 5, 8.)

```python
limit = int(input('Enter the limit for fibonacci series: '))
a,b = 0, 1
print ('Fibonacci series upto', limit , ':')
while a <= limit:
        print (a, end= ',')
        a,b = b, a+b
```
